```python
order_customer = orders.merge(
    customers,
    on="customer_id",
    how="inner"
)

order_customer.head()
```

```python
order_products = order_items.merge(products,on="product_id",how="inner")
order_products.head()
```

```python
order_details=order_customer.merge(order_products,on="order_id",how="inner")
order_details.head()
```

```python
order_details=order_details.merge(payments,on="order_id",how="inner")
order_details.head()
```

Sales & Revenue Analysis:
What is the total revenue?

```python
order_details["Revenue"]=order_details["quantity"]*order_details["selling_price"]
```

```python
total_revenue=order_details["Revenue"].sum()
```

```python
print(total_revenue)
```

Which product generates the highest revenue?

```python
product_revenue=order_details.groupby("product_name")["Revenue"].sum()
product_revenue=product_revenue.sort_values(ascending=False)
product_revenue.head(1)
```

Which category generates the highest revenue?

```python
category_revenue=order_details.groupby("category")["Revenue"].sum()
category_revenue = category_revenue.sort_values(ascending=False)

category_revenue.head(1)
```

Which city generates the highest revenue?

```python
city_revenue=order_details.groupby("city")["Revenue"].sum()
city_revenue=city_revenue.sort_values(ascending=False)
city_revenue.head(1)
```

Which customer segment generates the highest revenue?

```python
customersegments_revenue=order_details.groupby("customer_segment")["Revenue"].sum()
customersegments_revenue=customersegments_revenue.sort_values(ascending=False)
customersegments_revenue.head(1)
```

Which payment method generates the highest revenue?

```python
payments_revenue=order_details.groupby("payment_method")["Revenue"].sum()
payments_revenue=payments_revenue.sort_values(ascending=False)
payments_revenue.head(1)
```

Which month has the highest revenue?

```python
order_details["Month"]=order_details["order_date"].dt.to_period("M")
```

```python
Month_revenue=order_details.groupby("Month")["Revenue"].sum()
Month_revenue=Month_revenue.sort_values(ascending=False)
Month_revenue.head(1)
```

Which product has the highest quantity sold?

```python
product_quantity = order_details.groupby("product_name")["quantity"].sum()
product_quantity=product_quantity.sort_values(ascending=False)
product_quantity.head(1)
```

Which category has the highest quantity sold?

```python
category_quantity = order_details.groupby("category")["quantity"].sum()

category_quantity = category_quantity.sort_values(ascending=False)

category_quantity.head(1)
```

What is the average order value (AOV)?

```python
order_revenue = order_details.groupby("order_id")["Revenue"].sum()
AOV = order_revenue.mean()

print(AOV)
```

Who are the Top 5 Customers by Revenue?

```python
customer_revenue = order_details.groupby("customer_name")["Revenue"].sum()

customer_revenue.sort_values(ascending=False).head(5)
```

How many Repeat Customers are there?

```python
customer_orders = order_details.groupby("customer_id")["order_id"].nunique()

repeat_customers = (customer_orders > 1).sum()

print(repeat_customers)
```

What is the Average Number of Items per Order?

```python
order_quantity = order_details.groupby("order_id")["quantity"].sum()

average_items_per_order = order_quantity.mean()

print(average_items_per_order)
```

Which Product Generates the Highest Profit?

```python
order_details["Profit"] = (
    order_details["quantity"] *
    (order_details["selling_price"] - order_details["cost_price"])
)
```

```python
product_profit = order_details.groupby("product_name")["Profit"].sum()

product_profit.sort_values(ascending=False).head(1)
```

Which Category Generates the Highest Profit?

```python
category_profit = order_details.groupby("category")["Profit"].sum()

category_profit.sort_values(ascending=False).head(1)
```

```python
import pymysql
print("PyMySQL is working")
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:1234@localhost:3306/ecommerce_db"
)

print("Connected successfully")
```

```python
customers.to_sql("customers", con=engine, if_exists="replace", index=False)

products.to_sql("products", con=engine, if_exists="replace", index=False)

orders.to_sql("orders", con=engine, if_exists="replace", index=False)

order_items.to_sql("order_items", con=engine, if_exists="replace", index=False)

payments.to_sql("payments", con=engine, if_exists="replace", index=False)

print("All 5 tables uploaded successfully")
```
